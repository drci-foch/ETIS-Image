```python
import torchio as tio
import os
import pandas as pd
```

```python
swi_folder = "E:\\Data_Booster\\data_ETIS_1092\\SkullStripping\\SWI_Skullstripped_Canonical"
tof_folder = "E:\\Data_Booster\\data_ETIS_1092\\SkullStripping\\TOF3D_Skullstripped_Resampled"
label_folder = "E:\\Data_Booster\\data_ETIS_1092\\CanonicalNormalized_Images\\SWI_Skullstripped_asReference_ImagebyImage\\MASK"

swi_files = sorted([os.path.join(swi_folder, file) for file in os.listdir(swi_folder) if file.endswith(".nii.gz")])
tof_files = sorted([os.path.join(tof_folder, file) for file in os.listdir(tof_folder) if file.endswith(".nii.gz")])
label_files = sorted([os.path.join(label_folder, file) for file in os.listdir(label_folder) if file.endswith(".nii.gz")])
```

```python
metadata = []

for swi, tof, label in zip(swi_files, tof_files, label_files):
    swi_img = tio.ScalarImage(swi)
    tof_img = tio.ScalarImage(tof)
    label_img = tio.LabelMap(label)

    metadata.append({
        "SWI Number": "_".join(os.path.basename(swi).split("_")[:2]),
        "TOF3D Number": "_".join(os.path.basename(tof).split("_")[:2]),
        "Label Number": "_".join(os.path.basename(tof).split("_")[:2]),
        "SWI Spacing": swi_img.spacing,
        "TOF3D Spacing": tof_img.spacing,
        "Label Spacing": label_img.spacing,
        "SWI Shape": swi_img.shape,
        "TOF3D Shape": tof_img.shape,
        "Label Shape": label_img.shape,
        "SWI Direction": swi_img.affine[:3, :3].tolist(),
        "TOF3D Direction": tof_img.affine[:3, :3].tolist(),
        "Label Direction": label_img.affine[:3, :3].tolist()
    })

metadata_df = pd.DataFrame(metadata)
metadata_df.to_csv("D:\\data_ETIS_1092\\Training\\Checkpoints\\Metadata_check\\Skullstripped_1092\\image_metadata_corrected.csv", index=False)
```
